```python
%sql
CREATE TABLE IF NOT EXISTS project01_databricks01.silver.quarantine (
    table_name STRING,
    check_name STRING,
    column_name STRING,
    failed_record STRING,
    failure_reason STRING,
    processed_date TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.customers (
    customer_id STRING,
    customer_unique_id STRING,
    customer_zip_code_prefix INTEGER,
    customer_city STRING,
    customer_state STRING,
    processing_time TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.orders (
    order_id STRING,
    customer_id STRING,
    order_status STRING,
    order_purchase_timestamp TIMESTAMP,
    order_approved_at TIMESTAMP,
    order_delivered_carrier_date TIMESTAMP,
    order_delivered_customer_date TIMESTAMP,
    order_estimated_delivery_date TIMESTAMP,
    processing_time TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.products (
    product_id STRING,
    product_category_name STRING,
    product_name_lenght INTEGER,
    product_description_lenght INTEGER,
    product_photos_qty INTEGER,
    product_weight_g DOUBLE,
    product_length_cm DOUBLE,
    product_height_cm DOUBLE,
    product_width_cm DOUBLE,
    processing_time TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.sellers (
    seller_id STRING,
    seller_zip_code_prefix INTEGER,
    seller_city STRING,
    seller_state STRING,
    processing_time TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.reviews (
    review_id STRING,
    order_id STRING,
    review_score INTEGER,
    review_comment_title STRING,
    review_comment_message STRING,
    review_creation_date DATE,
    review_answer_timestamp TIMESTAMP,
    processing_time TIMESTAMP
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver.order_items (
    order_id STRING,
    order_item_id STRING,
    product_id STRING,
    seller_id STRING,
    shipping_limit_date TIMESTAMP,
    price DOUBLE,
    freight_value DOUBLE,
    processing_time TIMESTAMP
)
USING DELTA;
```

```python
%sql

CREATE TABLE IF NOT EXISTS silver.payments (
    order_id STRING,
    payment_sequential INTEGER,
    payment_type STRING,
    payment_installments INTEGER,
    payment_value DOUBLE,
    processing_time TIMESTAMP
)
USING DELTA;
```
